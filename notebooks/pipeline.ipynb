```python
import cv2
from collections import deque, Counter, defaultdict
from ultralytics import YOLO

# ---------------- config ----------------
VIDEO_PATH = r"C:\Users\gonth\Downloads\Video_Generation_From_Color_Boxes.mp4"
MODEL_PATH = r"C:\Users\gonth\Downloads\best.pt"
CONF = 0.35
COMMANDS = {0: "LEFT", 1: "RIGHT", 2: "FORWARD"}

def command_from_class(cls):
    return COMMANDS.get(int(cls), "NONE")

model = YOLO(MODEL_PATH)

def run_pass(window):
    """window=1 -> raw per-frame class command; window>1 -> majority vote."""
    cap = cv2.VideoCapture(VIDEO_PATH)
    if not cap.isOpened():
        raise RuntimeError(f"Could not open video: {VIDEO_PATH}")
    history = defaultdict(lambda: deque(maxlen=window))
    last_cmd, flips = {}, defaultdict(int)
    events = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        r = model.track(frame, persist=True, conf=CONF,
                        tracker="botsort.yaml", verbose=False)[0]
        b = r.boxes
        if b is None or b.id is None:
            continue
        ids = b.id.int().cpu().numpy()
        clss = b.cls.int().cpu().numpy()
        for tid, cls in zip(ids, clss):
            tid = int(tid)
            history[tid].append(int(cls))
            voted = Counter(history[tid]).most_common(1)[0][0]
            cmd = command_from_class(voted)
            events += 1
            if tid in last_cmd and last_cmd[tid] != cmd:
                flips[tid] += 1
            last_cmd[tid] = cmd
    cap.release()
    return dict(flips), sum(flips.values()), events

print("Running pass A: no smoothing (window=1) ...")
flips_raw, total_raw, events_raw = run_pass(1)
print("Running pass B: smoothing (window=8) ...")
flips_sm, total_sm, events_sm = run_pass(8)

all_ids = sorted(set(flips_raw) | set(flips_sm))
print("\n" + "=" * 46)
print(f"{'Track ID':>9} | {'flips raw':>10} | {'flips w=8':>10}")
print("-" * 46)
for tid in all_ids:
    print(f"{tid:>9} | {flips_raw.get(tid,0):>10} | {flips_sm.get(tid,0):>10}")
print("-" * 46)
print(f"{'TOTAL':>9} | {total_raw:>10} | {total_sm:>10}")
print("=" * 46)

reduction = (100 * (total_raw - total_sm) / total_raw) if total_raw else 0
print(f"\nCommand events (raw / w=8): {events_raw} / {events_sm}")
print(f"Flip reduction from smoothing: {reduction:.1f}%")
print(f"\nConclusion: majority-vote smoothing over 8 frames reduced per-track "
      f"command flips by {reduction:.1f}% ({total_raw} -> {total_sm}) "
      f"on {len(all_ids)} tracked objects.")
```

```python
import cv2
import numpy as np
from collections import deque, Counter, defaultdict
from ultralytics import YOLO

# ---------------- config ----------------
VIDEO_PATH = r"C:\Users\gonth\Downloads\Video_Generation_From_Color_Boxes.mp4"
MODEL_PATH = r"C:\Users\gonth\Downloads\best.pt"
CONF = 0.35
COMMANDS = {0: "LEFT", 1: "RIGHT", 2: "FORWARD"}

def command_from_class(cls):
    return COMMANDS.get(int(cls), "NONE")

model = YOLO(MODEL_PATH)

def iou(a, b):
    xa, ya = max(a[0], b[0]), max(a[1], b[1])
    xb, yb = min(a[2], b[2]), min(a[3], b[3])
    inter = max(0, xb - xa) * max(0, yb - ya)
    ua = (a[2]-a[0])*(a[3]-a[1]) + (b[2]-b[0])*(b[3]-b[1]) - inter
    return inter / ua if ua > 0 else 0

def metrics(seq_by_id):
    """seq_by_id: id -> list of commands in frame order.
    Returns total flips, total switches, mean purity."""
    flips = switches = 0
    purities = []
    for cmds in seq_by_id.values():
        if not cmds:
            continue
        flips += sum(1 for i in range(1, len(cmds)) if cmds[i] != cmds[i-1])
        switches += len(set(cmds)) - 1
        top = Counter(cmds).most_common(1)[0][1]
        purities.append(top / len(cmds))
    mean_purity = float(np.mean(purities)) if purities else 0.0
    return flips, switches, mean_purity

def run():
    cap = cv2.VideoCapture(VIDEO_PATH)
    if not cap.isOpened():
        raise RuntimeError(f"Could not open video: {VIDEO_PATH}")

    raw_seq   = defaultdict(list)   # pos-matched raw detection command
    track_seq = defaultdict(list)   # tracker class command
    vote_seq  = defaultdict(list)   # tracker + voting command
    hist = defaultdict(lambda: deque(maxlen=8))

    frame_id = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frame_id += 1

        # --- layer 1: raw detection (no tracker) ---
        det = model(frame, conf=CONF, verbose=False)[0]
        raw_boxes = []
        if det.boxes is not None and len(det.boxes):
            xyxy = det.boxes.xyxy.cpu().numpy()
            cls  = det.boxes.cls.int().cpu().numpy()
            raw_boxes = list(zip(xyxy, cls))

        # --- layers 2&3: tracked ---
        tr = model.track(frame, persist=True, conf=CONF,
                         tracker="botsort.yaml", verbose=False)[0]
        if tr.boxes is None or tr.boxes.id is None:
            continue
        t_xyxy = tr.boxes.xyxy.cpu().numpy()
        t_ids  = tr.boxes.id.int().cpu().numpy()
        t_cls  = tr.boxes.cls.int().cpu().numpy()

        for box, tid, cls in zip(t_xyxy, t_ids, t_cls):
            tid = int(tid)
            # layer 2: tracker class
            track_seq[tid].append(command_from_class(cls))
            # layer 3: voting
            hist[tid].append(int(cls))
            voted = Counter(hist[tid]).most_common(1)[0][0]
            vote_seq[tid].append(command_from_class(voted))
            # layer 1: match this track to nearest raw detection by IoU
            best, best_iou = None, 0.3
            for rbox, rcls in raw_boxes:
                v = iou(box, rbox)
                if v > best_iou:
                    best, best_iou = int(rcls), v
            if best is not None:
                raw_seq[tid].append(command_from_class(best))

    cap.release()
    return raw_seq, track_seq, vote_seq

raw_seq, track_seq, vote_seq = run()

rows = [
    ("1. Raw detection (no track)", metrics(raw_seq)),
    ("2. Tracked (BoT-SORT)",       metrics(track_seq)),
    ("3. Tracked + voting (w=8)",   metrics(vote_seq)),
]

print("=" * 66)
print(f"{'Layer':<30} | {'flips':>6} | {'switches':>8} | {'purity':>7}")
print("-" * 66)
for name, (f, s, p) in rows:
    print(f"{name:<30} | {f:>6} | {s:>8} | {p:>6.3f}")
print("=" * 66)
print("\npurity = mean fraction of a track's frames on its dominant command "
      "(1.000 = perfectly stable).")
print("Read across the three layers: where do flips/switches drop and purity "
      "rise? That shows which stage actually stabilizes the command.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# ---------------- config ----------------
VIDEO_PATH = r"C:\Users\gonth\Downloads\Video_Generation_From_Color_Boxes.mp4"
MODEL_PATH = r"C:\Users\gonth\Downloads\best.pt"
CONF = 0.35
COMMANDS = {0: "LEFT", 1: "RIGHT", 2: "FORWARD"}   # blue, red, white

def command_from_class(cls):
    return COMMANDS.get(int(cls), "NONE")

model = YOLO(MODEL_PATH)
cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    raise RuntimeError(f"Could not open video: {VIDEO_PATH}")

# ---------------- zone state ----------------
ZONE_Y = None            # horizontal trigger line (set on first frame)
fired_ids = set()        # IDs that already fired -> one command per object
prev_cy = {}             # last-seen center-y per track id, for crossing test
frame_id = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame_id += 1

    # set the sorting-zone line once, across the vertical middle of the frame
    if ZONE_Y is None:
        ZONE_Y = int(frame.shape[0] * 0.5)

    # tracker-native tracking: class travels WITH the identity (fixes classes[i] bug)
    results = model.track(frame, persist=True, conf=CONF,
                          tracker="botsort.yaml", verbose=False)[0]

    # draw the horizontal sorting zone (visible in screenshots)
    cv2.line(frame, (0, ZONE_Y), (frame.shape[1], ZONE_Y), (0, 0, 255), 2)
    cv2.putText(frame, "SORTING ZONE", (10, ZONE_Y - 8), 0, 0.6, (0, 0, 255), 2)

    boxes = results.boxes
    if boxes is not None and boxes.id is not None:
        xyxy = boxes.xyxy.cpu().numpy()
        ids  = boxes.id.int().cpu().numpy()
        clss = boxes.cls.int().cpu().numpy()

        for (x1, y1, x2, y2), tid, cls in zip(xyxy, ids, clss):
            tid = int(tid)
            cmd = command_from_class(cls)
            cy = int((y1 + y2) / 2)   # box center y

            # always draw the tracked box + label
            cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
            cv2.putText(frame, f"ID {tid} {cmd}", (int(x1), int(y1) - 5),
                        0, 0.7, (0, 255, 0), 2)

            # --- crossing test: did this box move across the line since last frame? ---
            crossed = False
            if tid in prev_cy:
                # true if the line sits between the previous and current center-y
                if (prev_cy[tid] - ZONE_Y) * (cy - ZONE_Y) <= 0:
                    crossed = True
            prev_cy[tid] = cy

            # fire the command ONCE, on crossing
            if crossed and tid not in fired_ids:
                fired_ids.add(tid)
                print(f"Frame {frame_id} -> BOX ID {tid} CROSSED ZONE -> {cmd}")
                cv2.putText(frame, f"FIRED: {cmd}", (int(x1), int(y1) - 28),
                            0, 0.7, (0, 0, 255), 2)

    img = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    plt.imshow(img)
    plt.axis("off")
    plt.pause(0.01)
    plt.clf()

cap.release()
print(f"\nDone. {len(fired_ids)} objects crossed the zone and were sorted.")
```
